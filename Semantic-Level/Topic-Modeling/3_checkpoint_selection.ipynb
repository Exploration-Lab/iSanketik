```python
import numpy as np
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.preprocessing import normalize
from umap import UMAP
from hdbscan import HDBSCAN

emb_best = np.load("./embeddings/embeddings_best.npy")
emb_ep35 = np.load("./embedding_epoch35/embeddings.npy")
emb_ep45 = np.load("./embedding_epoch45/embeddings.npy")
emb_ep60 = np.load("./embedding_epoch60/embeddings.npy")

def evaluate_embeddings(embeddings, name):
    emb = normalize(embeddings)

    umap = UMAP(
        n_neighbors=60,
        n_components=5,
        min_dist=0.0,
        metric="cosine",
        random_state=42
    )

    reduced = umap.fit_transform(emb)

    hdb = HDBSCAN(
        min_cluster_size=120,
        min_samples=30,
        metric="euclidean",
        cluster_selection_method="leaf",
        prediction_data=True
    )

    labels = hdb.fit_predict(reduced)

    mask = labels != -1
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    noise_pct = (labels == -1).sum() / len(labels) * 100

    if mask.sum() > 0 and n_clusters > 1:
        sil = silhouette_score(reduced[mask], labels[mask], sample_size=5000)
        db = davies_bouldin_score(reduced[mask], labels[mask])
    else:
        sil, db = -1, 999

    print(f"\n{'='*40}")
    print(f"Checkpoint: {name}")
    print(f"  Clusters found : {n_clusters}")
    print(f"  Noise points   : {noise_pct:.1f}%")
    print(f"  Silhouette     : {sil:.4f}")
    print(f"  Davies-Bouldin : {db:.4f}")
    print(f"{'='*40}")

    return {
        "name": name,
        "n_clusters": n_clusters,
        "noise_pct": noise_pct,
        "silhouette": sil,
        "davies_bouldin": db
    }

results = []
for emb, name in [
    (emb_best,  "best_backbone (ep94)"),
    (emb_ep35,  "epoch_35"),
    (emb_ep45,  "epoch_45"),
    (emb_ep60,  "epoch_60"),
]:
    results.append(evaluate_embeddings(emb, name))
```
